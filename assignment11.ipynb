```sql
-- Q1. Display customer_name, order_date, and today's date using CURRENT_DATE()
SELECT
    customer_name,
    order_date,
    CURRENT_DATE() AS today
FROM audio_sales;
```

```sql
-- Q2. Calculate the number of days between order_date and delivery_date using DATEDIFF()
SELECT
    customer_name,
    order_date,
    delivery_date,
    DATEDIFF(delivery_date, order_date) AS delivery_days
FROM audio_sales;
```

```sql
-- Q3. Find the date that is 7 days after each order_date using DATE_ADD()
SELECT
    customer_name,
    order_date,
    DATE_ADD(order_date, 7) AS date_after_7_days
FROM audio_sales;
```

```sql
-- Q4. Extract the year, month, and day from order_date using EXTRACT()
SELECT
    customer_name,
    order_date,
    EXTRACT(YEAR FROM order_date) AS year,
    EXTRACT(MONTH FROM order_date) AS month,
    EXTRACT(DAY FROM order_date) AS day
FROM audio_sales;
```

```sql
-- Q5. Create product_type using CASE WHEN
-- AirPods → Apple Product
-- Earphones → Other Brand
SELECT
    customer_name,
    product_name,
    category,
    CASE
        WHEN category = 'AirPods' THEN 'Apple Product'
        WHEN category = 'Earphones' THEN 'Other Brand'
    END AS product_type
FROM audio_sales;
```

```sql
-- Q6. Replace NULL values in discount with 0 using COALESCE()
SELECT
    customer_name,
    discount,
    COALESCE(discount, 0) AS final_discount
FROM audio_sales;
```

```sql
-- Q7. Replace NULL values in rating with 0 using COALESCE()
SELECT
    customer_name,
    rating,
    COALESCE(rating, 0) AS final_rating
FROM audio_sales;
```

```sql
-- Q8. Convert a rating of 0 into NULL using NULLIF()
SELECT
    customer_name,
    rating,
    NULLIF(rating, 0) AS rating_without_zero
FROM audio_sales;
```

```sql
-- Q9. Calculate discount percentage using NULLIF()
-- Formula: discount / price × 100
SELECT
    customer_name,
    price,
    discount,
    (COALESCE(discount, 0) / NULLIF(price, 0)) * 100 AS discount_percentage
FROM audio_sales;
```

```sql
-- Q10. Calculate delivery days and create a delivery category using CASE WHEN
-- 3 days or less → Fast
-- 4–5 days → Normal
-- More than 5 days → Slow
SELECT
    customer_name,
    order_date,
    delivery_date,
    DATEDIFF(delivery_date, order_date) AS delivery_days,
    CASE
        WHEN DATEDIFF(delivery_date, order_date) <= 3 THEN 'Fast'
        WHEN DATEDIFF(delivery_date, order_date) BETWEEN 4 AND 5 THEN 'Normal'
        ELSE 'Slow'
    END AS delivery_category
FROM audio_sales;
```

```sql
-- Q11. Display final discount, calculate sales, and classify sales using CASE WHEN
SELECT
    customer_name,
    product_name,
    COALESCE(discount, 0) AS final_discount,
    price * quantity AS sales,
    CASE
        WHEN price * quantity >= 500 THEN 'High Sales'
        WHEN price * quantity >= 200 THEN 'Medium Sales'
        ELSE 'Low Sales'
    END AS sales_category
FROM audio_sales;
```

```sql
-- Q12. Use at least 5 functions: CURRENT_DATE, DATEDIFF, DATE_ADD, EXTRACT, COALESCE, NULLIF
SELECT
    customer_name,
    order_date,
    CURRENT_DATE() AS today,
    DATEDIFF(delivery_date, order_date) AS delivery_days,
    DATE_ADD(order_date, 7) AS date_after_7_days,
    EXTRACT(YEAR FROM order_date) AS year,
    COALESCE(discount, 0) AS final_discount,
    NULLIF(rating, 0) AS rating_without_zero
FROM audio_sales;
```

```sql
-- Q13. Bonus: Use at least 5 functions in one query
SELECT
    customer_name,
    order_date,
    delivery_date,

    CURRENT_DATE() AS today,

    DATEDIFF(delivery_date, order_date) AS delivery_days,

    DATE_ADD(order_date, 7) AS date_after_7_days,

    EXTRACT(YEAR FROM order_date) AS order_year,

    COALESCE(discount, 0) AS final_discount,

    NULLIF(rating, 0) AS final_rating

FROM audio_sales;
```
