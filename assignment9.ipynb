```sql
-- ============================================
-- PART A: RANKING FUNCTIONS
-- ============================================

-- Q1. Rank all watches by price (descending) using RANK()

SELECT 
    model_name,
    price,
    RANK() OVER (ORDER BY price DESC) AS price_rank
FROM watch_sales;
```

```sql
-- Q2. Rank all watches by price using DENSE_RANK()

SELECT 
    model_name,
    price,
    DENSE_RANK() OVER (ORDER BY price DESC) AS price_rank
FROM watch_sales;
```

```sql
-- Q3. Assign unique sequential number to every sale by sale_date

SELECT 
    sale_id,
    model_name,
    sale_date,
    ROW_NUMBER() OVER (ORDER BY sale_date ASC) AS sale_number
FROM watch_sales;
```

```sql

-- Q4. Rank watches by units_sold within each category

SELECT 
    model_name,
    category,
    units_sold,
    RANK() OVER (
        PARTITION BY category 
        ORDER BY units_sold DESC
    ) AS category_rank
FROM watch_sales;
```

```sql
-- Q5. Top 3 best-selling watches in each category using DENSE_RANK()

WITH ranked_watches AS
(
    SELECT 
        model_name,
        category,
        units_sold,
        DENSE_RANK() OVER (
            PARTITION BY category
            ORDER BY units_sold DESC
        ) AS sales_rank
    FROM watch_sales
)
SELECT 
    model_name,
    category,
    units_sold,
    sales_rank
FROM ranked_watches
WHERE sales_rank <= 3;
```

```sql
-- Q6. Most recent sale for each brand using ROW_NUMBER()

WITH recent_sales AS
(
    SELECT 
        brand_name,
        model_name,
        sale_date,
        ROW_NUMBER() OVER (
            PARTITION BY TRIM(brand_name)
            ORDER BY sale_date DESC
        ) AS rn
    FROM watch_sales
)
SELECT 
    brand_name,
    model_name,
    sale_date
FROM recent_sales
WHERE rn = 1;
```

```sql
-- ============================================
-- PART B: STRING FUNCTIONS
-- ============================================

-- Q8. Create full_name using CONCAT()

SELECT 
    CONCAT(customer_first_name, ' ', customer_last_name) AS full_name
FROM watch_sales;
```

```sql
-- Q9. Remove leading and trailing spaces from brand_name

SELECT 
    TRIM(brand_name) AS clean_brand_name
FROM watch_sales;
```

```sql
-- Q10. Extract first 4 characters of model_name

SELECT 
    model_name,
    SUBSTRING(model_name, 1, 4) AS first_four_characters
FROM watch_sales;
```

```sql
-- Q11. Replace hyphens with spaces

SELECT 
    model_name,
    REPLACE(model_name, '-', ' ') AS new_model_name
FROM watch_sales;
```

```sql
-- Q12. Convert country to uppercase

SELECT 
    country,
    UPPER(country) AS uppercase_country
FROM watch_sales;
SELECT 
    country,
    LOWER(country) AS lowercase_country
FROM watch_sales;
```

```sql
-- Q13. Combine TRIM() and UPPER() to clean brand_name

SELECT 
    brand_name,
    UPPER(TRIM(brand_name)) AS clean_brand_name
FROM watch_sales;
```

```sql
-- ============================================
-- PART C: MIXED / COMBINED PRACTICE
-- ============================================

-- Q14. Clean names, create full_name,
-- and rank customers by units_sold

WITH customer_data AS
(
    SELECT
        CONCAT(
            UPPER(SUBSTRING(TRIM(customer_first_name), 1, 1)),
            LOWER(SUBSTRING(TRIM(customer_first_name), 2)),
            ' ',
            UPPER(SUBSTRING(TRIM(customer_last_name), 1, 1)),
            LOWER(SUBSTRING(TRIM(customer_last_name), 2))
        ) AS full_name,
        units_sold
    FROM watch_sales
)
SELECT
    full_name,
    units_sold,
    DENSE_RANK() OVER (ORDER BY units_sold DESC) AS units_rank
FROM customer_data;
```

```sql
-- Q15. Mask last name and number alphabetically

SELECT
    ROW_NUMBER() OVER (
        ORDER BY customer_last_name
    ) AS row_num,

    CONCAT(
        SUBSTRING(customer_last_name, 1, 1),
        REPEAT('*', LENGTH(customer_last_name) - 1)
    ) AS masked_last_name

FROM watch_sales
ORDER BY customer_last_name;
```
