```sql
-- Create Table
CREATE TABLE MarineInsurance
(
    PolicyID INT,
    CustomerName VARCHAR(50),
    VesselName VARCHAR(50),
    CargoType VARCHAR(50),
    PolicyAmount DECIMAL(12,2),
    ClaimAmount DECIMAL(12,2),
    Premium DECIMAL(10,2),
    Port VARCHAR(50),
    PolicyStatus VARCHAR(20),
    PolicyDate DATE
);
```

```sql
-- Q1: Policies with premium greater than average premium

WITH AvgPremium AS
(
    SELECT AVG(Premium) AS Avg_Premium
    FROM MarineInsurance
)

SELECT *
FROM MarineInsurance
WHERE Premium > (SELECT Avg_Premium FROM AvgPremium);
```

```sql
-- Q2: Customer(s) having the highest policy amount

WITH MaxPolicy AS
(
    SELECT MAX(PolicyAmount) AS Max_Policy
    FROM MarineInsurance
)

SELECT CustomerName, PolicyAmount
FROM MarineInsurance
WHERE PolicyAmount = (SELECT Max_Policy FROM MaxPolicy);
```

```sql
-- Q3: Customer(s) having the highest claim amount

WITH MaxClaim AS
(
    SELECT MAX(ClaimAmount) AS Max_Claim
    FROM MarineInsurance
)

SELECT CustomerName, ClaimAmount
FROM MarineInsurance
WHERE ClaimAmount = (SELECT Max_Claim FROM MaxClaim);
```

```sql
-- Q4: Total premium collected from each port

WITH PortPremium AS
(
    SELECT Port,
           SUM(Premium) AS Total_Premium
    FROM MarineInsurance
    GROUP BY Port
)

SELECT *
FROM PortPremium;
```

```sql
-- Q5: Ports having total premium greater than 15000

WITH PortPremium AS
(
    SELECT Port,
           SUM(Premium) AS Total_Premium
    FROM MarineInsurance
    GROUP BY Port
)

SELECT *
FROM PortPremium
WHERE Total_Premium > 15000;
```

```sql
-- Q6: Total policy amount for each cargo type

WITH CargoPolicy AS
(
    SELECT CargoType,
           SUM(PolicyAmount) AS Total_Policy
    FROM MarineInsurance
    GROUP BY CargoType
)

SELECT *
FROM CargoPolicy;
```

```sql
-- Q7: Average policy amount for each port

WITH AvgPolicy AS
(
    SELECT Port,
           AVG(PolicyAmount) AS Avg_Policy
    FROM MarineInsurance
    GROUP BY Port
)

SELECT *
FROM AvgPolicy;
     
```

```sql
-- Q8: Policies with policy amount greater than overall average

WITH AvgPolicy AS
(
    SELECT AVG(PolicyAmount) AS Avg_Policy
    FROM MarineInsurance
)

SELECT *
FROM MarineInsurance
WHERE PolicyAmount > (SELECT Avg_Policy FROM AvgPolicy);
```

```sql
-- Q9: Port having the highest total policy amount

WITH PortPolicy AS
(
    SELECT Port,
           SUM(PolicyAmount) AS Total_Policy
    FROM MarineInsurance
    GROUP BY Port
),
MaxPolicy AS
(
    SELECT MAX(Total_Policy) AS Max_Total
    FROM PortPolicy
)

SELECT *
FROM PortPolicy
WHERE Total_Policy = (SELECT Max_Total FROM MaxPolicy);
```

```sql
-- Q10: Policies with claim amount greater than average claim amount

WITH AvgClaim AS
(
    SELECT AVG(ClaimAmount) AS Avg_Claim
    FROM MarineInsurance
)

SELECT *
FROM MarineInsurance
WHERE ClaimAmount > (SELECT Avg_Claim FROM AvgClaim);
```

```sql
-- Q11: Policies having policy amount greater than their port average

WITH PortAverage AS
(
    SELECT Port,
           AVG(PolicyAmount) AS Avg_Policy
    FROM MarineInsurance
    GROUP BY Port
)

SELECT M.*
FROM MarineInsurance M
JOIN PortAverage P
ON M.Port = P.Port
WHERE M.PolicyAmount > P.Avg_Policy;
```

```sql
-- Q12: Cargo types with total claim amount greater than 100000

WITH CargoClaim AS
(
    SELECT CargoType,
           SUM(ClaimAmount) AS Total_Claim
    FROM MarineInsurance
    GROUP BY CargoType
)

SELECT *
FROM CargoClaim
WHERE Total_Claim > 100000;
```

```sql
-- Q13: Customers paying premium greater than their port average

WITH PortAverage AS
(
    SELECT Port,
           AVG(Premium) AS Avg_Premium
    FROM MarineInsurance
    GROUP BY Port
)

SELECT M.CustomerName,
       M.Port,
       M.Premium
FROM MarineInsurance M
JOIN PortAverage P
ON M.Port = P.Port
WHERE M.Premium > P.Avg_Premium;
```

```sql
-- Q14: Policies contributing more than 25% of their port's total premium

WITH PortPremium AS
(
    SELECT Port,
           SUM(Premium) AS Total_Premium
    FROM MarineInsurance
    GROUP BY Port
)

SELECT M.PolicyID,
       M.CustomerName,
       M.Port,
       M.Premium
FROM MarineInsurance M
JOIN PortPremium P
ON M.Port = P.Port
WHERE M.Premium > (0.25 * P.Total_Premium);
```

```sql
-- Q15: Total premium, total claim amount, and total policy amount for each port

WITH PortSummary AS
(
    SELECT Port,
           SUM(Premium) AS Total_Premium,
           SUM(ClaimAmount) AS Total_Claim,
           SUM(PolicyAmount) AS Total_Policy
    FROM MarineInsurance
    GROUP BY Port
)

SELECT *
FROM PortSummary;
```
