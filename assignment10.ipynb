```sql
-- =========================================================
-- SALES DATA
-- =========================================================

CREATE TABLE gold_sales (
    Order_ID INT,
    Customer_ID INT,
    Customer_Name VARCHAR(50),
    Department VARCHAR(30),
    Order_Date DATE,
    Sales_Amount DECIMAL(10,2)
);
```

```sql
-- 1. Calculate total sales for each Department
SELECT
    Order_ID,
    Customer_Name,
    Department,
    Sales_Amount,
    SUM(Sales_Amount) OVER (
        PARTITION BY Department
    ) AS Department_Total_Sales

FROM gold_sales;
```

```sql
-- 2. Calculate average sales for each Department
SELECT
    Order_ID,
    Customer_Name,
    Department,
    Sales_Amount,
    AVG(Sales_Amount) OVER (
        PARTITION BY Department
    ) AS Department_Average_Sales

FROM gold_sales;
```

```sql
-- 3. Calculate total sales for the entire company
SELECT
    Order_ID,
    Customer_Name,
    Sales_Amount,
    SUM(Sales_Amount) OVER () AS Company_Total_Sales

FROM gold_sales;
```

```sql
-- 4. Calculate average sales for the entire company
SELECT
    Order_ID,
    Customer_Name,
    Sales_Amount,
    AVG(Sales_Amount) OVER () AS Company_Average_Sales

FROM gold_sales;
```

```sql
-- 5. Calculate running total of sales by date
SELECT
    Order_ID,
    Customer_Name,
    Order_Date,
    Sales_Amount,
    SUM(Sales_Amount) OVER (
        ORDER BY Order_Date, Order_ID
    ) AS Running_Total

FROM gold_sales;
```

```sql
-- 6. Calculate running total separately for each Department
SELECT
    Order_ID,
    Customer_Name,
    Department,
    Order_Date,
    Sales_Amount,
    SUM(Sales_Amount) OVER (
        PARTITION BY Department
        ORDER BY Order_Date, Order_ID
    ) AS Department_Running_Total

FROM gold_sales;
```

```sql
-- 7. Find the highest-selling customer in each Department

WITH RankedSales AS
(
    SELECT
        Order_ID,
        Customer_Name,
        Department,
        Sales_Amount,
        ROW_NUMBER() OVER (
            PARTITION BY Department
            ORDER BY Sales_Amount DESC
        ) AS Row_Num

    FROM gold_sales
)
SELECT
    Order_ID,
    Customer_Name,
    Department,
    Sales_Amount
FROM RankedSales
WHERE Row_Num = 1;
```

```sql
-- 8. Find the top 2 sales levels in each Department

WITH RankedSales AS
(
    SELECT
        Order_ID,
        Customer_Name,
        Department,
        Sales_Amount,
        DENSE_RANK() OVER (
            PARTITION BY Department
            ORDER BY Sales_Amount DESC
        ) AS Sales_Rank

    FROM gold_sales
)
SELECT
    Order_ID,
    Customer_Name,
    Department,
    Sales_Amount,
    Sales_Rank
FROM RankedSales
WHERE Sales_Rank <= 2
ORDER BY Department, Sales_Rank;
```
